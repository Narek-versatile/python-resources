```python
class single:
    _instance = None

    def __new__(cls, num):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance 

    def __init__(self, num):
        self.num = num


x = single(1)
y = single(1)
z = single(2)
print(x.num)
print(y.num)
print(z.num)
```
